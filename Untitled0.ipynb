```python
# Konfiguracja danych użytkownika Git
!git config --global user.email "student@uczelnia.edu.pl"
!git config --global user.name "Bartosz Sobecki"

# Klonowanie repozytorium MEDICA
!git clone https://github.com/<username>/MEDICA.git
%cd MEDICA

# Przykładowy moduł – zarządzanie pacjentami (lorem ipsum)
class Patient:
    def __init__(self, name, pesel, doctor):
        self.name = name
        self.pesel = pesel
        self.doctor = doctor
        self.visits = []

    def add_visit(self, date, description):
        self.visits.append({'date': date, 'desc': description})

    def get_summary(self):
        return f"Pacjent: {self.name}, PESEL: {self.pesel}, "
               f"Lekarz: {self.doctor}, Wizyty: {len(self.visits)}"

# Przykładowe dane testowe
p1 = Patient("Jan Kowalski", "85010112345", "Dr Anna Nowak")
p1.add_visit("15.04.2026", "Badanie kontrolne")
p1.add_visit("10.03.2026", "Konsultacja kardiologiczna")
print(p1.get_summary())

# Zapis zmian i push do GitHub
!git add .
!git commit -m "Add: modul zarzadzania pacjentami"
!git push origin main
```
